```python
input = [2, 3, 5, 4, 6, 7, 8]
# output = 4   # [4, 6, 7, 8]
```

```python
def find_longest_sequence(nums):
    """Length of longest contiguous strictly increasing segment."""
    if not nums:
        return 0
    start = 0
    best = 1
    for i in range(1, len(nums)):
        if nums[i] > nums[i - 1]:
            best = max(best, i - start + 1)
        else:
            start = i
    return best
```

```python
# Quick checks (run after editing find_longest_sequence)
assert find_longest_sequence([2, 3, 5, 4, 6, 7, 8]) == 4
assert find_longest_sequence([5, 4, 3]) == 1
assert find_longest_sequence([]) == 0
print("All assertions passed")
```
